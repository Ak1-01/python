```python
# 3 sum

class Solution:
    def threeSum(self, nums: List[int]) -> List[List[int]]:

        res = []
        nums.sort()

        length = len(nums)
        for i in range(length-2):
            if i >0 and nums[i] == nums[i -1]:
                continue
            l = i +1
            r = length -1

            while l <r:
                total = nums[i] + nums[l] + nums[r]

                if total < 0:
                    l = l+1
                elif total >0:
                    r = r-1
                else:
                    res.append([nums[i] , nums[l] , nums[r]])
                    while l <r and nums[l] == nums[l+1]:
                        l +=1
                    while l <r and nums[r] == nums[r-1]:
                        r -=1
                    l +=1
                    r-=1
        return res

#Algo - sort array then take 2 pointer l and r l = i +1 r = last
#total =[ i + i +1 or l + r]
"""total >0
r =  r-1
total <0
l = l+1


total =0     
l + l+1       
r = r-1                    
arr.append([nums[i] , nums[l], nums[r]]
"""
```

```python
# Inheritance when one child class inherit some behaviour from parent class along with its own attributes
  
class animal():
    def __init__(self):
        print("Animal is created")
    def name(self):
        print("animal")
    def do(self):
        print("runs")

class dog(animal):
    def __init__(self):
        super().__init__()
        print("dog is created")
    def name(self):
        print("dog")
    def does(self):
        print("barks")

d = dog()
d.__init__()
d.name()
d.does()
d.do()
```

```python
from os import *
from sys import *
from collections import *
from math import *

def pairSum(arr, s):
    n = len(arr)
    ans= []
    for i in range(n):
        for j in range(i +1,n):
            if (arr[i] +arr[j] ==s):
                pair[-1, for i in range(2)]
                pair[0] = arr[i]
                pair[1] = arr[j]
                ans.append(pair)

    res = [[-1 for j in range(2) for i in range(len(ans))]]
    for i in range(len[ans]):
        a = ans[i][0]
        b = ans[j][1]

        res[i][0] = min[a,b]
        res[j][1] = max[a,b]
    res = sorted(res,key : lambda x: x[0])
    return res
```

```python
#Reverse a string


def push(stack, i):
    stack.append(i)

def IsEmpty(stack):
    if (len(stack) == 0):
        return True
    return False

def pop(stack):
    if IsEmpty(stack):
        return
    return stack.pop()

def reverse(str):
    n = len(str)
    stack = []
    for i in range(n):
        push(stack,str[i])
    str = ""
    for i in range(n):
        str += pop(stack)
    return str
    

str = "HELLO"
print("Reversed String - ", reverse(str))
reverse(str)

        
    
```

```python
#TRIE - used for efficient search of words
#TC - o(n*L) n is size of list l is length of word 
#Hashmap avg(o(n*l)) Worst- o(n*) with trie TC - o(l) tc for insertion o(l)



class TrieNode:
    def __init__(self):
        self.child = {}
        self.IsEnd = False

class Trie:
    def __init__(self):
        self.node = TrieNode()
    
    def insert(self,string):
        n = len(string)
        if n ==0:
            self.node.IsEnd = True
            
        else:
            if string[0] in self.node.child:
                self.node.child[string[0]].insert(string[1:])
            else:
                self.node.child[string[0]] = Trie()
                self.node.child[string[0]].insert(string[1:])
            return
    def search(self,word):
        temp = self.node
        for i in range(len(word)):
            if word[i] not in temp.child:
                return False
            else:
                temp = temp.child[word[i]].node
                
        if (temp.IsEnd == True):
            return True
        else:
            return False

root = Trie()
            
root.insert("beds")
root.insert("bedroom")
root.insert("bed")
print(root.search("bed"))
        
```
