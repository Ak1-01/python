```python
# 3 sum

class Solution:
    def threeSum(self, nums: List[int]) -> List[List[int]]:

        res = []
        nums.sort()

        length = len(nums)
        for i in range(length-2):
            if i >0 and nums[i] == nums[i -1]:
                continue
            l = i +1
            r = length -1

            while l <r:
                total = nums[i] + nums[l] + nums[r]

                if total < 0:
                    l = l+1
                elif total >0:
                    r = r-1
                else:
                    res.append([nums[i] , nums[l] , nums[r]])
                    while l <r and nums[l] == nums[l+1]:
                        l +=1
                    while l <r and nums[r] == nums[r-1]:
                        r -=1
                    l +=1
                    r-=1
        return res

#Algo - sort array then take 2 pointer l and r l = i +1 r = last
#total =[ i + i +1 or l + r]
"""total >0
r =  r-1
total <0
l = l+1


total =0     
l + l+1       
r = r-1                    
arr.append([nums[i] , nums[l], nums[r]]
"""
```

```python
#Roman to INteger


lass Solution:
    def romanToInt(self, s: str) -> int:
        dic = { 
            'I' :            1,
            'V' :           5,
            'X'  :          10,
            'L'   :         50,
            'C'    :        100,
            'D'     :       500,
            'M'     :      1000,
        }
        total = 0
        prev =0
        curr = 0 
        for i in range(len(s)):
            curr = dic[s[i]]
            if curr > prev:
                total = total +  curr - 2*prev
            else:
                total = total + curr
            prev = curr
        return total


        
A#make a dic - two pointer prev and curr and total =0 loop to string curr = 1st element of string if curr <prev then total = total + curr -2*prev else total += curr and then prev = curr return total
```

```python

#Reorder data in log files


class Solution:
    def reorderLogFiles(self, logs: List[str]) -> List[str]:
        res1,res2 = [],[]
        for log in logs:
            if(log.split()[1]).isdigit():
                res2.append(log)
            else:
                res1.append(log.split())
        res1.sort(key = lambda x : x[0])
        res1.sort(key = lambda x: x[1:])

        for i in range(len(res1)):
            res1[i] = (" ".join(res1[i]))
        res1.extend(res2)
        return res1


#Take two arrays in 1 store letter logs  in lexical order in another store digitlogs  then combine both and return in form of string      
        
```

```python
Maximum subarray

class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        total_sum = max_sum = nums[0]

        for i in nums[1:]:
            total_sum = max(i, total_sum +i)
            max_sum = max(max_sum , total_sum)
        return max_sum

        
#check 1st index its max then check next now find max (sum till 1st or prev index and the curr index ) and update it
#[-2,1,-3,4,-1,2,1,-5,4] at index 0 its index 0 at index  1 =1 or -2 _1 so 2 ,at index 2- either i +- 3 or -3 so -2 ,at index 3 -2 +4 or 4 so 4
```

```python
longest common prefix


class Solution:
    def longestCommonPrefix(self, strs: List[str]) -> str:

        if len(strs) == 0:
            return ""
        minlen =  len(strs[0])
        for i in range(len(strs)):
            minlen = min(len(strs[i]), minlen)
        lcp = ""
        i = 0
        while i <minlen:
            char = strs[0][i]
            for j in range(1,len(strs)):
                if strs[j][i] != char:
                    return lcp
            lcp =  lcp + char
            i +=1
        return lcp
#Find smallest string then length coz longest common cant be greater than smlllest substring
#to calc min string - for  loop minlen variable store len of oth index string and compare to every index and update if len(minstring) in smaller than len(currstring)
#i = 0 lcp = empty string if no commo prefix " "while i < minlen char = 1st element of 1st ist str[0][]1] for k in range(1, len(str))compare if str 1sr letter != char return lcp else lcp = lcp +char i +=1
```

```python
#Subtree of another tree

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def isSubtree(self, s:TreeNode,t:TreeNode) -> bool:
        if s is None and t is None:
            return True
        if t is None:
            return True
        if s is None and t is not None:
            return False
        return self.issame(s,t) or self.isSubtree(s.left,t) or self.isSubtree(s.right,t)

    

    def issame(self,s,t):
        if s is None and t is None:
            return True
        if s is None or t is None:
            return False
        return s.val ==t.val and self.issame(s.left,t.left) and self.issame(s.right,t.right)

# check subtrees if matches then True make issame function then edge cases like if tree is empty 

```

```python
#Happy Number 202
class Solution:
    def isHappy(self, n: int) -> bool:
        def sqsum(num):
            result = 0
            while num >0:
                r = num % 10
                result = result + r*r #19%10 -  9
                num = num //10   #To devide the number like 19 - 1 and 9
            return result
        seem = set()
        while sqsum(n) not in seem:
            sum1 = sqsum(n)
            if sum1 == 1:
                return True
            else:
                seem.add(sum1)
                n =sum1
        return False

```
