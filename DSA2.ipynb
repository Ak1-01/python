```python
# Set Matrix zeros or 1



from typing import List

def setZeros(matrix: List[List[int]]) -> None:
    row =len(matrix)
    col = len(matrix[0])


    row_zero = [False for _ in range(row)]
    col_zero = [False for _ in range(col)]

    for i in range(row):
        for j in range(col):

            if (matrix[i][j] ==0): #for 1 - reolace 0 to 1

                row_zero[i] = True
                col_zero[j] = True
    

    for i in range (row):
        for j in range(col):
            if row_zero[i] or col_zero[j] :
                matrix[i][j] = 0 #for 1 replace 0 to 1

    return matrix
```

```python
Remove Duplicates from Unsorted LL

def removeDuplicates(head):
    prev = None
    curr = head
    map = {}

    while curr != None:
        map[curr.data] = False
        curr = curr.next
    
    curr = head
    while curr!= None:
        val = curr.data
        if map[val] == True:
            prev.next = curr.next
            del curr
            curr = prev.next
        
        else:
            map[val] = True
            prev = curr
            curr = curr.next
    return head
```

```python
Odd and even positioned linked list


def oddEvenLinkedList(head):
    if (head == None or head.next == None):
        return head

    oddNode = head
    evenNode= head.next
    currNode = evenNode.next
    evenHead = evenNode

    pos = 1
    while (currNode != None):
        if (pos & 1):
            oddNode.next = currNode
            oddNode = currNode
        else:
            evenNode.next = currNode
            evenNode = currNode

        pos +=1
        currNode = currNode.next

    oddNode.next = evenHead
    evenNode.next = None

    return head
```
